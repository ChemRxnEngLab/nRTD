```python
import sympy as sp
import numpy as np
```

```python
s, t = sp.symbols("s t")
tau_p, tau_a, tau_m = sp.symbols("tau_p tau_a tau_m", real=True, positive=True)
beta = sp.symbols("beta", real=True, positive=True)

G = sp.exp(-tau_p * s) / (1 + beta + tau_a * s - (beta / (1 + tau_m)))

display(G)
```

```python
E = sp.inverse_laplace_transform(G, s, t).doit()
display(E)
```

```python
fn = sp.lambdify(
    [t, tau_p, tau_a, tau_m, beta],
    E,
    modules=["numpy"],
)
fn(1, 1, 1, 1, 1)
```
