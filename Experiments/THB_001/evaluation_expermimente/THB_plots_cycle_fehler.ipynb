```python
# Titel für das gesamte Diagramm
diagram_title = "alle cycle"

# --- FIGURE SETUP  ---
fig = plt.figure(figsize=(16 * ICIW_Plots.cm2inch, 13 * ICIW_Plots.cm2inch))

ax = ICIW_Plots.make_square_subplots(
    fig=fig,
    ax_width=7 * ICIW_Plots.cm2inch, 
    ax_layout=(1, 1),
    h_sep=2 * ICIW_Plots.cm2inch,
    v_sep=1 * ICIW_Plots.cm2inch,
    sharex=False,
    sharey=False,
    sharelabel=False,
    xlabel=[r'$t\,/\,\mathrm{s}$'],
    ylabel=[r'Molenbruch $x_{Ar}$ / [-]']
)

# --- LOOP ÜBER DIE LISTE ---
for data_obj, label, color in plot_configs_0:
    t = data_obj['time']
    y_avg = data_obj['avg']
    y_upper = data_obj['upper']
    y_lower = data_obj['lower']

    # Statistischen Fehlerbereich plotten 
    ax[0, 0].fill_between(
        t,
        y1=y_upper,
        y2=y_lower,
        alpha=0.2,    # Transparentfaktor
        color=color,
        edgecolor='none'
    )

    # Hauptlinie plotten
    ax[0, 0].plot(
        t,
        y_avg,
        color=color,
        label=label,
        linewidth=1
    )

# ---  FINISH & DESIGN ---
ax[0, 0].set_title(diagram_title, fontsize=12, fontweight='bold')
ax[0, 0].grid(True, linestyle='--', alpha=0.5)

# Legende 
ax[0, 0].legend(
    loc='upper left', 
    bbox_to_anchor=(1.02, 1), 
    borderaxespad=0, 
    fontsize='small',
    frameon=True
)

plt.tight_layout()
plt.show()

```

```python
diagram_title = "alle cycle / normiert"

# --- FIGURE SETUP (2 Zeilen, unten 2 Spalten) ---
fig = plt.figure(figsize=(16 * ICIW_Plots.cm2inch, 18 * ICIW_Plots.cm2inch))

# Wir nutzen ein 2x2 Layout
# ax_layout=(rows, cols)
ax = ICIW_Plots.make_square_subplots(
    fig=fig,
    ax_width=6 * ICIW_Plots.cm2inch, 
    ax_layout=(2, 2),           
    h_sep=2.5 * ICIW_Plots.cm2inch,
    v_sep=2.5 * ICIW_Plots.cm2inch,
    sharex=False,
    sharey=False,
    sharelabel=False,
    xlabel=r'$t\,/\,\mathrm{s}$', 
    ylabel=r'Molenbruch $x_{Ar}$ / [-]'
)

ax_main = ax[0, 0]   # Gesamtübersicht 
ax_zoom1 = ax[1, 0]  # Zoom Anstieg (unten links)
ax_zoom2 = ax[1, 1]  # Zoom Abfall (unten rechts)

if ax.shape == (2, 2):
    ax[0, 1].axis('off')

# --- 3. LOOP ÜBER DIE LISTE ---
for data_obj, label, color in plot_configs_0:
    t = data_obj['time']
    y_avg = data_obj['avg']
    y_upper = data_obj['upper']
    y_lower = data_obj['lower']

    # --- NORMIERUNG ---
    mask = (t >= 70) & (t <= 90)
    norm_factor = np.mean(y_avg[mask]) if np.any(mask) else 1.0

    y_avg_n = y_avg / norm_factor
    y_upper_n = y_upper / norm_factor
    y_lower_n = y_lower / norm_factor

    # --- PLOTTEN IN DIE GEZIELTEN ACHSEN ---
    for target_ax in [ax_main, ax_zoom1, ax_zoom2]:
        target_ax.fill_between(t, y_upper_n, y_lower_n, alpha=0.15, color=color, edgecolor='none')
        target_ax.plot(t, y_avg_n, color=color, label=label, linewidth=1.2)

# --- 4. FINISH & DESIGN ---

# Plot 1: Gesamtübersicht
ax_main.set_title(f"{diagram_title}\nGesamtübersicht", fontsize=11, fontweight='bold')

# Plot 2: Zoom Anstieg
ax_zoom1.set_xlim(3, 10)
ax_zoom1.set_title("Zoom: Anstieg (0 - 30 s)", fontsize=10, fontweight='bold')

# Plot 3: Zoom Abfall
ax_zoom2.set_xlim(120, 130)
ax_zoom2.set_title("Zoom: Abfall (120 - 140 s)", fontsize=10, fontweight='bold')

# Globales Design für die aktiven Achsen
for target_ax in [ax_main, ax_zoom1, ax_zoom2]:
    target_ax.grid(True, linestyle='--', alpha=0.4)
    target_ax.set_ylim(-0.05, 1.15) 

# Legende nur oben rechts neben dem Hauptplot
ax_zoom2.legend(loc='center left', bbox_to_anchor=(1.05, 1), 
               borderaxespad=0, fontsize='small', frameon=True)

plt.tight_layout()
plt.show()
```
