```python


DEFAULTS = dict(
    ylim=(-0.05, 1.15),
    norm_window=(70, 90),   # plateau window used for the normalisation
    n_markers=8,            # markers per panel, not per curve
    marker_size=4.0,
    line_width=1.2,
    band_alpha=0.15,
    legend_ncol=2,
    ax_width_cm=6.0,
    h_sep_cm=1.6,
    fig_size_cm=(16, 9),
    xlabel=r'$t\,/\,\mathrm{s}$',
    ylabel=r'$x_{Ar}$ / 1',
)


def _marker_every(t, xlim, n_markers):

    idx = np.flatnonzero((t >= xlim[0]) & (t <= xlim[1]))
    if idx.size == 0:
        return None
    step = max(1, idx.size // max(n_markers, 1))
    return (int(idx[0] % step), step)


def plot_zooms(configs, xlim_rise, xlim_fall, legend_entries=None, **kw):
    """Two zoom panels, shared y scale, manual legend.

    configs         list of (data, label, colour, marker, linestyle)
    xlim_rise       x limits of the left panel, e.g. (3, 10)
    xlim_fall       x limits of the right panel, e.g. (120, 130)
    legend_entries  optional override, list of (colour, marker, linestyle, label);
                    None = taken from configs, which cannot drift out of sync
    """
    o = {**DEFAULTS, **kw}

    fig = plt.figure(figsize=(o['fig_size_cm'][0] * ICIW_Plots.cm2inch,
                              o['fig_size_cm'][1] * ICIW_Plots.cm2inch))

    ax = ICIW_Plots.make_square_subplots(
        fig=fig,
        ax_width=o['ax_width_cm'] * ICIW_Plots.cm2inch,
        ax_layout=(1, 2),
        h_sep=o['h_sep_cm'] * ICIW_Plots.cm2inch,
        sharex=False,
        sharey=True,               # same scale, so both flanks stay comparable
        sharelabel=False,
        xlabel=o['xlabel'],
        ylabel=o['ylabel'],
    )

    ax = np.atleast_1d(ax).ravel()
    ax_rise, ax_fall = ax[0], ax[1]
    panels = ((ax_rise, xlim_rise), (ax_fall, xlim_fall))

    for data, label, color, marker, ls in configs:
        t = data['time']
        y_avg, y_upper, y_lower = data['avg'], data['upper'], data['lower']

        mask = (t >= o['norm_window'][0]) & (t <= o['norm_window'][1])
        norm_factor = np.mean(y_avg[mask]) if np.any(mask) else 1.0

        y_avg_n = y_avg / norm_factor
        y_upper_n = y_upper / norm_factor
        y_lower_n = y_lower / norm_factor

        # no label= here: the legend is assembled by hand below
        for target_ax, xlim in panels:
            target_ax.fill_between(t, y_upper_n, y_lower_n,
                                   alpha=o['band_alpha'], color=color, edgecolor='none')
            target_ax.plot(t, y_avg_n,
                           color=color, linestyle=ls, linewidth=o['line_width'],
                           marker=marker, markersize=o['marker_size'],
                           markerfacecolor='none',   # open symbols keep the line readable
                           markeredgewidth=0.9,
                           markevery=_marker_every(t, xlim, o['n_markers']))

    ax_rise.set_xlim(*xlim_rise)
    ax_fall.set_xlim(*xlim_fall)

    for target_ax, _ in panels:
        target_ax.grid(True, linestyle='--', alpha=0.4)
        target_ax.set_ylim(*o['ylim'])
        target_ax.set_axisbelow(True)

    # y label and y tick labels only on the left panel
    ax_fall.set_ylabel('')
    ax_fall.tick_params(labelleft=False)

    entries = (legend_entries if legend_entries is not None else
               [(c, m, ls, lab) for _, lab, c, m, ls in configs])

    handles = [Line2D([0], [0], color=col, marker=mk, linestyle=ls,
                      linewidth=o['line_width'], markersize=o['marker_size'],
                      markerfacecolor='none', markeredgewidth=0.9, label=lab)
               for col, mk, ls, lab in entries]

    fig.legend(handles=handles, loc='center left', bbox_to_anchor=(1.02, 0.5),
               ncol=1, frameon=False, fontsize='small',
               bbox_transform=ax_fall.transAxes)

    return fig, (ax_rise, ax_fall)
```

```python
T_SWITCH_BU = 0.0            # start of the build-up half period, in s
T_SWITCH_BT = 120.0          # start of the back-transient half period, in s
 
BASE_WINDOW_BU = (0.0, 2.5)      # signal before the rise starts
PLATEAU_WINDOW = (70.0, 90.0)    # plateau of the BU phase (same as the plot)
BASE_WINDOW_BT = (190.0, 230.0)  # plateau of the BT phase
 
SEARCH_RISE = (0.0, 60.0)        # window in which the rise is looked for
SEARCH_FALL = (120.0, 180.0)
 
LEVELS = (0.10, 0.50, 0.90)
# -----------------------------------------------------------------------------
 
 
def _window_mean(t, y, window):
    m = (t >= window[0]) & (t <= window[1])
    if not np.any(m):
        raise ValueError(f'no data points inside the window {window}')
    return float(np.mean(y[m]))
 
 
def _crossing(t, y, level, rising):

    hit = np.flatnonzero(y >= level) if rising else np.flatnonzero(y <= level)
    if hit.size == 0:
        return np.nan
    i = int(hit[0])
    if i == 0:
        return float(t[0])
    y0, y1 = float(y[i - 1]), float(y[i])
    if y1 == y0:
        return float(t[i])
    return float(t[i - 1] + (level - y0) * (t[i] - t[i - 1]) / (y1 - y0))
 
 
def _flank_levels(t, y, search, base, top, rising, levels=LEVELS):
    """Normalise one flank to 0..1 and return the crossing times."""
    m = (t >= search[0]) & (t <= search[1])
    ts, ys = t[m], y[m]
    span = top - base
    if abs(span) < 1e-12:
        return {lv: np.nan for lv in levels}
    yn = (ys - base) / span                     # 1 at the plateau, 0 at the baseline
    return {lv: _crossing(ts, yn, lv, rising) for lv in levels}
 
 
def flank_metrics(configs, levels=LEVELS, verbose=True, return_rows=True):
    """t(x = 0.5) and t10-90 for both flanks, for every entry of a config list.
 
    configs   the same (data, label, colour, marker, linestyle) lists that are
              used for the plots, so figure and table cannot diverge
    """
    rows = []
 
    for entry in configs:
        data, label = entry[0], entry[1]
        t = np.asarray(data['time'], float)
        y = np.asarray(data['avg'], float)
 
        base_bu = _window_mean(t, y, BASE_WINDOW_BU)
        plateau = _window_mean(t, y, PLATEAU_WINDOW)
        base_bt = _window_mean(t, y, BASE_WINDOW_BT)
 
        rise = _flank_levels(t, y, SEARCH_RISE, base_bu, plateau, True, levels)
        fall = _flank_levels(t, y, SEARCH_FALL, base_bt, plateau, False, levels)
 
        # uncertainty from the confidence band that is already in the data
        t50_rise_band, t50_fall_band = [], []
        for key in ('upper', 'lower'):
            if key not in data:
                continue
            yb = np.asarray(data[key], float)
            b_bu = _window_mean(t, yb, BASE_WINDOW_BU)
            plt_b = _window_mean(t, yb, PLATEAU_WINDOW)
            b_bt = _window_mean(t, yb, BASE_WINDOW_BT)
            t50_rise_band.append(
                _flank_levels(t, yb, SEARCH_RISE, b_bu, plt_b, True, (0.5,))[0.5])
            t50_fall_band.append(
                _flank_levels(t, yb, SEARCH_FALL, b_bt, plt_b, False, (0.5,))[0.5])
 
        def _halfspread(v):
            v = [x for x in v if np.isfinite(x)]
            return (max(v) - min(v)) / 2 if len(v) >= 2 else np.nan
 
        rows.append({
            'label': label,
            # medians, relative to the respective switching event
            't50_rise': rise[0.50] - T_SWITCH_BU,
            't50_fall': fall[0.50] - T_SWITCH_BT,
            # widths: both reported as a positive duration
            't10_90_rise': rise[0.90] - rise[0.10],
            't90_10_fall': fall[0.10] - fall[0.90],
            # uncertainty of the median from the confidence band
            'u_t50_rise': _halfspread(t50_rise_band),
            'u_t50_fall': _halfspread(t50_fall_band),
            # plateau levels, as a plausibility check on the normalisation
            'base_bu': base_bu, 'plateau': plateau, 'base_bt': base_bt,
        })
 
    if verbose:
        print('=' * 104)
        print('Flank metrics   (all times in s, relative to the switching event)')
        print('=' * 104)
        print(f'{"configuration":32s} {"t50 rise":>10s} {"+/-":>7s} '
              f'{"t50 fall":>10s} {"+/-":>7s} {"t10-90 rise":>13s} {"t90-10 fall":>13s}')
        print('-' * 104)
        for r in rows:
            print(f'{str(r["label"])[:32]:32s} '
                  f'{r["t50_rise"]:10.2f} {r["u_t50_rise"]:7.2f} '
                  f'{r["t50_fall"]:10.2f} {r["u_t50_fall"]:7.2f} '
                  f'{r["t10_90_rise"]:13.2f} {r["t90_10_fall"]:13.2f}')
        print('-' * 104)
 
        # differences against the first entry: that is what goes into the text
        ref = rows[0]
        print()
        print(f'Differences against "{ref["label"]}"')
        print(f'{"configuration":32s} {"d t50 rise":>12s} {"d t50 fall":>12s} '
              f'{"d width rise":>14s} {"d width fall":>14s}')
        print('-' * 104)
        for r in rows[1:]:
            print(f'{str(r["label"])[:32]:32s} '
                  f'{r["t50_rise"] - ref["t50_rise"]:12.2f} '
                  f'{r["t50_fall"] - ref["t50_fall"]:12.2f} '
                  f'{r["t10_90_rise"] - ref["t10_90_rise"]:14.2f} '
                  f'{r["t90_10_fall"] - ref["t90_10_fall"]:14.2f}')
        print('-' * 104)
        print('A shift in t50 at constant width is a dead-volume effect.')
        print('A change in width at constant t50 is a dispersion effect.')
        print('A difference smaller than the +/- column is not resolvable.')
        print()
 
        # normalisation check: the baselines should be flat and the plateau stable
        bad = [r for r in rows
               if abs(r['base_bu']) > 0.05 * abs(r['plateau'])
               or abs(r['base_bt']) > 0.05 * abs(r['plateau'])]
        if bad:
            print('WARNING - baseline not close to zero for: '
                  + ', '.join(str(r['label']) for r in bad))
            print('Check BASE_WINDOW_BU / BASE_WINDOW_BT and the switching times.')
            print()
 
    return rows if return_rows else None
 
 

 
```

```python
MOM_WINDOW_RISE = (0.0, 115.0)    # integration window of the rising flank, in s
MOM_WINDOW_FALL = (120.0, 235.0)  # same for the falling flank
 
TAIL_EXTRAPOLATE = False   # True = fit an exponential tail and add its contribution
TAIL_FIT_FRAC = 0.25       # last fraction of the window used for that fit
# -----------------------------------------------------------------------------
 
_trapz_np = getattr(np, 'trapezoid', None) or np.trapz
 
 
def _step_to_F(t, y, window, base, top, rising):
    """Cut out one flank and normalise it to a cumulative distribution 0..1.
 
    For the falling flank the signal is mirrored, so that F rises from 0 to 1
    in both cases and the same moment formulas apply.
    """
    m = (t >= window[0]) & (t <= window[1])
    ts = np.asarray(t[m], float) - window[0]
    ys = np.asarray(y[m], float)
 
    span = top - base
    if abs(span) < 1e-12:
        raise ValueError('plateau and baseline coincide')
 
    F = (ys - base) / span if rising else 1.0 - (ys - base) / span
    return ts, np.clip(F, 0.0, None)
 
 
def _moments_from_F(t, F, tail_extrapolate=TAIL_EXTRAPOLATE):
    """tbar and sigma^2 from the cumulative distribution, plus truncation info."""
    g = 1.0 - F                      # survival function
 
    tbar = float(_trapz_np(g, t))
    m2 = 2.0 * float(_trapz_np(t * g, t))
    var = m2 - tbar ** 2
 
    # how much of tbar comes from the last 20 % of the window
    cut = t[0] + 0.8 * (t[-1] - t[0])
    late = t >= cut
    tail_frac = (float(_trapz_np(g[late], t[late])) / tbar) if tbar > 0 else np.nan
    F_end = float(F[-1])
 
    if tail_extrapolate:
        # g ~ A exp(-t/theta) on the last part of the window
        n = max(5, int(TAIL_FIT_FRAC * t.size))
        tt, gg = t[-n:], g[-n:]
        ok = gg > 1e-6
        if ok.sum() >= 3:
            sl, ic = np.polyfit(tt[ok], np.log(gg[ok]), 1)
            if sl < 0:
                theta = -1.0 / sl
                A = np.exp(ic)
                t_end = float(t[-1])
                g_end = A * np.exp(-t_end / theta)
                tbar += g_end * theta
                m2 += 2.0 * g_end * theta * (t_end + theta)
                var = m2 - tbar ** 2
 
    return {'tbar': tbar, 'var': float(var),
            'sigma': float(np.sqrt(max(var, 0.0))),
            'tail_frac': tail_frac, 'F_end': F_end}
 
 
def exact_moments(configs, verbose=True, tail_extrapolate=TAIL_EXTRAPOLATE):
    """tbar and sigma of both flanks, computed from the measured F curves."""
    rows = []
 
    for entry in configs:
        data, label = entry[0], entry[1]
        t = np.asarray(data['time'], float)
        y = np.asarray(data['avg'], float)
 
        base_bu = _window_mean(t, y, BASE_WINDOW_BU)
        plateau = _window_mean(t, y, PLATEAU_WINDOW)
        base_bt = _window_mean(t, y, BASE_WINDOW_BT)
 
        tr, Fr = _step_to_F(t, y, MOM_WINDOW_RISE, base_bu, plateau, True)
        tf, Ff = _step_to_F(t, y, MOM_WINDOW_FALL, base_bt, plateau, False)
 
        mr = _moments_from_F(tr, Fr, tail_extrapolate)
        mf = _moments_from_F(tf, Ff, tail_extrapolate)
 
        # The baseline window must lie entirely BEFORE the breakthrough.
        # If it already contains part of the rise, the baseline is too high,
        # F starts negative, gets clipped, and tbar comes out too large.
        y_at_base_end = float(np.interp(BASE_WINDOW_BU[1], t, y))
        f_at_base_end = (y_at_base_end - base_bu) / (plateau - base_bu)
 
        rows.append({'label': label,
                     'F_at_base_end': f_at_base_end,
                     'tbar_rise': mr['tbar'], 'sigma_rise': mr['sigma'],
                     'var_rise': mr['var'],
                     'tbar_fall': mf['tbar'], 'sigma_fall': mf['sigma'],
                     'var_fall': mf['var'],
                     'tail_rise': mr['tail_frac'], 'tail_fall': mf['tail_frac'],
                     'Fend_rise': mr['F_end'], 'Fend_fall': mf['F_end']})
 
    if verbose:
        note = '  [WITH exponential tail extrapolation]' if tail_extrapolate else ''
        print('=' * 108)
        print('Exact moments from the measured F curves' + note)
        print('=' * 108)
        print(f'{"configuration":32s} {"tbar rise":>10s} {"sigma":>8s} {"var":>9s} '
              f'{"tbar fall":>10s} {"sigma":>8s} {"var":>9s} {"tail R":>7s} {"tail F":>7s}')
        print('-' * 108)
        for r in rows:
            print(f'{str(r["label"])[:32]:32s} '
                  f'{r["tbar_rise"]:10.2f} {r["sigma_rise"]:8.2f} {r["var_rise"]:9.2f} '
                  f'{r["tbar_fall"]:10.2f} {r["sigma_fall"]:8.2f} {r["var_fall"]:9.2f} '
                  f'{r["tail_rise"] * 100:6.1f}% {r["tail_fall"] * 100:6.1f}%')
        print('-' * 108)
        print('tbar and sigma in s, var in s^2. tail = share of tbar from the last')
        print('20 % of the integration window; above ~5 % the window is too short')
        print('and sigma^2 is underestimated.')
 
        ref = rows[0]
        print()
        print(f'Differences against "{ref["label"]}"   (variances are additive in series)')
        print(f'{"configuration":32s} {"d tbar R":>10s} {"d var R":>10s} '
              f'{"d tbar F":>10s} {"d var F":>10s}')
        print('-' * 108)
        for r in rows[1:]:
            print(f'{str(r["label"])[:32]:32s} '
                  f'{r["tbar_rise"] - ref["tbar_rise"]:10.2f} '
                  f'{r["var_rise"] - ref["var_rise"]:10.2f} '
                  f'{r["tbar_fall"] - ref["tbar_fall"]:10.2f} '
                  f'{r["var_fall"] - ref["var_fall"]:10.2f}')
        print('-' * 108)
 
        bad = [r for r in rows if r['Fend_rise'] < 0.98 or r['Fend_fall'] < 0.98]
        if bad:
            print()
            print('WARNING - F does not reach 1 within the window for: '
                  + ', '.join(str(r['label']) for r in bad))
            print('Widen MOM_WINDOW_RISE / MOM_WINDOW_FALL or check the plateau levels.')
 
        early = [r for r in rows if r['F_at_base_end'] > 0.02]
        if early:
            print()
            print('WARNING - the breakthrough starts inside BASE_WINDOW_BU for: '
                  + ', '.join(f'{r["label"]} (F={r["F_at_base_end"]:.3f})' for r in early))
            print('The baseline is then too high and tbar of the rising flank comes')
            print('out too large. Shorten BASE_WINDOW_BU.')
        print()
 
    return rows
 
 
def compare_to_kernels(rows, pairs, kernel_moments=None):
    """Differences between two configurations against the nRTD kernel moments.
 
    pairs            list of (label_a, label_b, description); the difference
                     b - a isolates the compartments that b contains in
                     addition to a
    kernel_moments   optional dict {description: (tbar, var)} from the nRTD run
    """
    by_label = {str(r['label']): r for r in rows}
 
    print('=' * 108)
    print('Difference between two configurations   ->   isolated compartment(s)')
    print('=' * 108)
    print(f'{"isolates":34s} {"d tbar R":>10s} {"d var R":>10s} '
          f'{"d tbar F":>10s} {"d var F":>10s}   nRTD')
    print('-' * 108)
    for a, b, desc in pairs:
        if a not in by_label or b not in by_label:
            print(f'{desc:34s}   labels not found: {a!r} / {b!r}')
            continue
        ra, rb = by_label[a], by_label[b]
        ref = ''
        if kernel_moments and desc in kernel_moments:
            kt, kv = kernel_moments[desc]
            ref = f'  tbar {kt:.2f} s, var {kv:.2f} s^2'
        print(f'{desc:34s} '
              f'{rb["tbar_rise"] - ra["tbar_rise"]:10.2f} '
              f'{rb["var_rise"] - ra["var_rise"]:10.2f} '
              f'{rb["tbar_fall"] - ra["tbar_fall"]:10.2f} '
              f'{rb["var_fall"] - ra["var_fall"]:10.2f}{ref}')
    print('-' * 108)
    print('Valid only where the two configurations differ by compartments in')
    print('series and in nothing else. A and B do NOT qualify: A contains the')
    print('connection tubing E4prime, B the inlet piping E1, so their difference')
    print('is E4prime against E1 and not an added compartment.')
    print()
 
 
# =============================================================================
#  EVALUATION
# =============================================================================
print('### PLOT 1  measurement positions')
mom_positions = exact_moments(configs_positions)
 

 
```

```python
def flow_scaling(rows, flows, keys=('t50_rise', 't50_fall',
                                    't10_90_rise', 't90_10_fall'),
                 label=''):

    x = 1.0 / np.asarray(flows, float)
    out = {}
    print(f'--- {label} ---')
    for key in keys:
        y = np.array([r[key] for r in rows], float)
        slope, icpt = np.polyfit(x, y, 1)
        pred = slope * x + icpt
        ss_res = float(np.sum((y - pred) ** 2))
        ss_tot = float(np.sum((y - y.mean()) ** 2))
        r2 = 1 - ss_res / ss_tot if ss_tot > 0 else np.nan
        out[key] = (icpt, slope, r2)
        print(f'{key:14s} intercept {icpt:6.2f} s   slope {slope:7.1f}   '
              f'R2 {r2:6.3f}   at {max(flows):.0f}: {slope / max(flows):.2f} s')
        print('               residuals: ' +
              '  '.join(f'{f:.0f}: {v - p:+.3f}' for f, v, p in zip(flows, y, pred)))
    return out


def split_contributions(out_a, out_b, flow_ref, name_a='A', name_b='B'):
    """Difference between two configurations, split into the two parts.

    Both configurations share the capillary, so its contribution cancels here.
    """
    print(f'--- {name_a} minus {name_b} ---')
    print(f'{"quantity":14s} {"flow-independent":>18s} {"flow-dependent":>16s}')
    for key in out_a:
        if key not in out_b:
            continue
        d_i = out_a[key][0] - out_b[key][0]
        d_s = (out_a[key][1] - out_b[key][1]) / flow_ref
        print(f'{key:14s} {d_i:15.2f} s {d_s:13.2f} s')
    print('A large flow-independent part is NOT a dead volume in the main stream.')


FLOWS = [250, 125, 62.5]        # must match the order of the config list
out_cap = flow_scaling(rows_capillary, FLOWS, label='A  capillary')
out_vor = flow_scaling(rows_vorreactor, FLOWS, label='B  inlet piping')
split_contributions(out_cap, out_vor, flow_ref=250)
```
