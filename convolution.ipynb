```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 10, 100)

# inlet concentration
c_t = np.zeros_like(t)
c_t[70:] = 1
c_t[30:70] = np.linspace(0, 1, 40)
```

```python
gauss = (
    lambda x, mu, sigma: 1
    / np.sqrt(2 * np.pi * sigma**2)
    * np.exp(-1 / 2 * ((x - mu) / sigma) ** 2)
)
E_t = gauss(t, 3.0, 0.5)
```

```python
# calculate the new time scale
t_c = np.linspace(2 * t[0], 2 * t[-1], 199)
```

```python
# convolve E with c
c_out = np.convolve(E_t, c_t)
```

```python
# plotting
plt.plot(t, c_t, label="c_in(t)")
plt.plot(t, E_t)
plt.plot(t_c, c_out)
plt.legend()
plt.show()
```
