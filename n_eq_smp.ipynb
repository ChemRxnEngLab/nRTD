```python
import sympy as smp
```

```python
# n0, n1, n2, nE, t0, t1, t2, tE = smp.symbols("n_0 n_1 n_2 n_E t_0 t_1 t_2 t_E")
n0, n1, n2, nE1, nE2 = smp.symbols("n_0 n_1 n_2 n_E1 n_E2", positive=True, real=True)
t0, t1, t2, tE1, tE2 = smp.symbols("t_0 t_1 t_2 t_E1, t_E2", positive=True, real=True)
```

```python
equations = [
    t0 - t1 + tE1,
    n0 - n1 + nE1 + 1,
    n0 / t0 - nE1 / tE1,
    t1 - t2 + tE2,
    n1 - n2 + nE2 + 1,
    nE2 * t1 - n1 * tE2,
]
```

```python
for equation in equations:
    display(equation)
```

```python
solution_set = smp.solve(equations[:3], (n0, n1, t0), set=True)
display(solution_set)
```

```python
solution_set = smp.nonlinsolve(equations[:3], (n0, n1, t0))
display(solution_set)
```

```python
solution_set = smp.solve(
    equations,
    n0,
    n1,
    # n2,
    # nE1,
    nE2,
    # t0,
    t1,
    t2,
    # tE1,
    tE2,
    dict=True,
)
solution = solution_set[0]
sub_dict = {
    # n0: 377,
    # n1,
    n2: 500,
    nE1: 122,
    # nE2,
    t0: 31,
    # t1,
    # t2,
    tE1: 20,
    # tE2,
}
for key, value in solution.items():
    print(key)

    value = value.subs(sub_dict)
    sub_dict.update({key: smp.N(value)})

print(sub_dict)
```
